1.-Programa que convierte grados celsius a farenheint
  Rocha Vazquez Jarvis Gael 1G

```python
def celsius_to_fahrenheit(celsius):
    fahrenheit = (celsius * 9/5) + 32
    return fahrenheit

# Get input from the user
try:
    celsius_input = float(input("Coloca la temperatura en grados celsius"))
    fahrenheit_output = celsius_to_fahrenheit(celsius_input)
    print(f"{celsius_input}°C es igual a {fahrenheit_output}°F")
except ValueError:
    print("numeros ilegibles. Porfavor coloca numeros legibles e intenta de nuevo.")
```

2.-Programa que convierte kilometros a millas

```python
def kilometers_to_miles(kilometers):
  miles = kilometers * 0.621371
  return miles

# Get input from the user
try:
  kilometers_input = float(input("Coloca la distancia en kilometros: "))
  miles_output = kilometers_to_miles(kilometers_input)
  print(f"{kilometers_input} km es igual a {miles_output} millas")
except ValueError:
  print("Distancia invalida. Porfavor ingresa un valor numerico de nuevo")
```

3.-Programa de calculo de propina

```python
#Get input from the user for the bill amount and tip percentage
try:
    bill_amount = float(input("Ingresa el total de la cuenta de la cena: "))
    tip_percentage = float(input("Ingresa el porcentaje de propina que deseas dejar (ejemplo. 30 para 30%): "))

    #Calculate the tip amount
    tip_amount = bill_amount * (tip_percentage / 100)

    #Calculate the total amount including tip
    total_amount = bill_amount + tip_amount


    #Display results
    print(f"\n Total de  la cuenta: ${bill_amount:.2f}")
    print(f"Porcentaje de propina: {tip_percentage}%")
    print(f"Cantidad de propina: ${tip_amount:.2f}")
    print(f"Total a pagar (incluyendo propina): ${total_amount:.2f}")

except ValueError:
    print("Invalido. porfavor ingresa solo numeros para la cuenta y el porcentaje que desees dar como propina")
except Exception as e:
    print(f"Ocurrio un error: {e}")
```
